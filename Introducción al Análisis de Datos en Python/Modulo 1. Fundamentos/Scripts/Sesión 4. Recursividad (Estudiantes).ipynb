## Formato Estándar (PEP 257)

````python
def suma(a: int, b: int) -> int:
    """
    Calcula la suma de dos números enteros.

    Args:
        a (int): Primer número.
        b (int): Segundo número.

    Returns:
        int: Resultado de la suma.
    """
    return a + b
````

````python
?suma
````

````python
def division(a: float, b: float) -> float:
    """
    Divide dos números.

    Parameters
    ----------
    a : float
        El numerador.
    b : float
        El denominador (no debe ser 0).

    Returns
    -------
    float
        Resultado de la división.

    Raises
    ------
    ValueError
        Si `b` es 0.
    """
    if b == 0:
        raise ValueError("El denominador no puede ser 0")
    return a / b
````

# Recursividad o recursión
> La recursividad es un tipo de funciones que se llaman a sí mismas para resolver un problema dividiéndolo en partes más sencillas.

Este tipo de funciones son muy conocidas, no solo por programadores sino también por **Santa Claus**.

## Santa Claus y la recursividad 🎅🏻
Todo niño se ha preguntado alguna vez: ¿cómo hace Santa para recorrer el mundo en una sola noche?

Una posible explicación es que Santa sigue una lista de casas en orden. Llega a una casa, deja los regalos, come las galletas y la leche, y luego continúa con la siguiente. Esta forma para resolver el problema se asemeja a un `for loop` y en la lógica de algoritmos se conoce como un algoritmo iterativo.

Sin embargo, a su edad, Santa no podría hacer todas las entregas solo. Por eso, la recursividad es una explicación más plausible sobre como se hacen las entregas desde el Polo Norte:
1. Santa le asigna la lista de todos los hogares al duende _chief operating officer_ (COO).
2. El duende COO divide la lista en dos y asigna la mitad de las casas a cada uno de sus dos duendes VP.
3. La lista de hogares sigue dividiéndose en la cadena de mando hasta que cada duende trabajador recibe un solo hogar para repartir los regalos.

En Python, esta estrategia se vería de la siguiente manera

````python
# Lista de casas con su respectivo regalo [)
casas = [
    ("Casa de Ana", "🎁"),
    ("Casa de Juan", "🧸"),
    ("Casa de María", "🎮"),
    ("Casa de Pedro", "📚"),
    ("Casa de Luis", "🦾")
]
````

````python
for i in casas:
    print("Para la", i[0], "se entrega el regalo", i[1])
````

````python
def entregar_regalos_recursivamente(casas):

    if len(casas) == 1:
        casa, regalo = casas[0]
        print("Para la", casa, "se entrega el regalo", regalo)

    else:  
        # Dividir el trabajo:
        mitad = len(casas) // 2
        primera_mitad = casas[:mitad]
        segunda_mitad = casas[mitad:]

        # CLAVE DE LA RECURSIVIDAD
        # Sub Duende 1
        entregar_regalos_recursivamente(primera_mitad)
        # Sub Duende 2
        entregar_regalos_recursivamente(segunda_mitad)
````

````python
entregar_regalos_recursivamente(casas)
````

Ahora que entendimos la intuición detrás de la recursividad, entendamos mejor su estructura.

Básicamente una función recursiva se llama a sí misma y repetirá su comportamiento **hasta que se cumpla una condición** para devolver un resultado. Todas las funciones recursivas comparten una estructura común compuesta por dos partes: el caso base y el caso recursivo.

```python
def funcion_recursiva(parametros):

    # Caso Base
    if condicion_de_parada:
        return resultado_base

    # Caso Recursivo
    else:
        return funcion_recursiva(parametros_modificados)
```
**Caso Base y Caso Recursivo**
- **Caso Base:** Esta es la condición bajo la cual la recursión se detiene. Es crucial para evitar bucles infinitos y garantizar que cada llamada recursiva reduzca el problema de alguna manera. En el ejemplo de Santa el Caso Base corresponde a que el número de hogares a repartir sea solo uno y entonces haya que ir a entregar ese regalo.
- **Caso Recursivo:** Es la parte de la función que incluye la llamada a sí misma. Debe conducir eventualmente al caso base. En el ejemplo de Santa el Caso Recursivo corresponde a dividir los hogares en mitades.

Para ilustrar nuestra lógica veamos el caso de sumar los números de una lista usando recursión y usando `for loops`.

````python
def suma_loops(lista):
    suma_total = 0
    for elemento in lista:
        suma_total += elemento
    return suma_total
````

````python
def suma_recursiva(lista):
    # Caso base: Si la lista tiene un solo elemento, se devuelve ese elemento
    if len(lista) == 1:
        return lista[0]
    else:
        # Se suma el primer elemento con la suma del resto de la lista
        return lista[0] + suma_recursiva(lista[1:])
````

````python
lista = [10, -2, 34, 78, 23, 11, -76, 132, 1]

suma_loops(lista)
````
