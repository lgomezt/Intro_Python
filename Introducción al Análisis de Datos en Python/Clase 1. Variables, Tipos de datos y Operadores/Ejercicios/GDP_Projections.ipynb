## Economic Concepts

### 1. Compound Growth Formula
When an economy grows at a constant rate each year, the future value is calculated as:
```
future_value = present_value × (1 + growth_rate) ^ years
```
This is **exponential growth**, not linear!

### 2. Rule of 70
A quick way to estimate doubling time for any growth rate:
```
doubling_time = 70 / (growth_rate × 100)
```
For example, at 3% growth: 70 / 3 ≈ 23.3 years to double.

## Important Python Concepts

### Exponentiation Operator
- In Python, use `**` for powers (NOT `^` or `*`)
- Example: `2 ** 3` equals `8`
- Warning: `^` in Python is XOR (bitwise), not exponentiation!

### Operator Precedence
- `70 / 0.03 * 100` evaluates left-to-right: `(70 / 0.03) * 100`
- Use parentheses to enforce order: `70 / (0.03 * 100)`
