## ¿Qué es la Recursividad?

> La recursividad es un tipo de funciones que se llaman a sí mismas para resolver un problema dividiéndolo en partes más sencillas.

En muchos problemas, la solución natural consiste en expresar el problema en términos de una versión más pequeña de sí mismo. La recursión nos permite hacer exactamente eso: escribir funciones que se invocan a sí mismas con datos cada vez más simples, hasta llegar a un caso tan pequeño que la respuesta es trivial.

Este tipo de funciones son muy conocidas, no solo por programadores sino también por **Santa Claus**.

## Santa Claus y la Recursividad 🎅🏻

Todo niño se ha preguntado alguna vez: ¿cómo hace Santa para recorrer el mundo en una sola noche?

Una posible explicación es que Santa sigue una lista de casas en orden. Llega a una casa, deja los regalos, come las galletas y la leche, y luego continúa con la siguiente. Esta forma para resolver el problema se asemeja a un `for loop` y en la lógica de algoritmos se conoce como un algoritmo iterativo.

Sin embargo, a su edad, Santa no podría hacer todas las entregas solo. Por eso, la recursividad es una explicación más plausible sobre como se hacen las entregas desde el Polo Norte:
1. Santa le asigna la lista de todos los hogares al duende _chief operating officer_ (COO).
2. El duende COO divide la lista en dos y asigna la mitad de las casas a cada uno de sus dos duendes VP.
3. La lista de hogares sigue dividiéndose en la cadena de mando hasta que cada duende trabajador recibe un solo hogar para repartir los regalos.

En Python, esta estrategia se vería de la siguiente manera:

````python
# Lista de casas con su respectivo regalo
casas = [
    ("Casa de Ana", "🎁"),
    ("Casa de Juan", "🧸"),
    ("Casa de María", "🎮"),
    ("Casa de Pedro", "📚")
]
````

````python
# Función recursiva para entregar los regalos
def entregar_regalos_recursivamente(casas):

    # Duende trabajador entregando el regalo
    if len(casas) == 1:
        casa, regalo = casas[0]
        print(f"Entregando {regalo} en {casa}")

    # Duende supervisor dividiendo el trabajo
    else:
        mitad = len(casas) // 2
        primera_mitad = casas[:mitad]
        segunda_mitad = casas[mitad:]

        # Divide el trabajo entre dos duendes
        entregar_regalos_recursivamente(primera_mitad)
        entregar_regalos_recursivamente(segunda_mitad)
````

````python
entregar_regalos_recursivamente(casas)
````

## Estructura de una Función Recursiva

Ahora que entendimos la intuición detrás de la recursividad, entendamos mejor su estructura.

Básicamente una función recursiva se llama a sí misma y repetirá su comportamiento **hasta que se cumpla una condición** para devolver un resultado. Todas las funciones recursivas comparten una estructura común compuesta por dos partes: el caso base y el caso recursivo.

```python
def funcion_recursiva(parametros):

    # Caso Base
    if condicion_de_parada:
        return resultado_base

    # Caso Recursivo
    else:
        return funcion_recursiva(parametros_modificados)
```

### Caso Base y Caso Recursivo

- **Caso Base:** Esta es la condición bajo la cual la recursión se detiene. Es crucial para evitar bucles infinitos y garantizar que cada llamada recursiva reduzca el problema de alguna manera. En el ejemplo de Santa el Caso Base corresponde a que el número de hogares a repartir sea solo uno y entonces haya que ir a entregar ese regalo.
- **Caso Recursivo:** Es la parte de la función que incluye la llamada a sí misma. Debe conducir eventualmente al caso base. En el ejemplo de Santa el Caso Recursivo corresponde a dividir los hogares en mitades.

### ¿Cómo piensa la computadora?

Cuando una función se llama a sí misma, la computadora "apila" cada llamada en una estructura llamada **call stack** (pila de llamadas). Cada llamada espera a que la siguiente termine para poder usar su resultado. Cuando se alcanza el caso base, las llamadas empiezan a "desapilarse" y devolver sus resultados en orden inverso.

## Ejemplo 1: Suma de una Lista

Para ilustrar nuestra lógica veamos el caso de sumar los números de una lista usando recursión y usando `for loops`.

````python
def suma_loops(lista):
    suma_total = 0
    for elemento in lista:
        suma_total += elemento
    return suma_total
````

````python
def suma_recursiva(lista):
    # Caso base: Si la lista tiene un solo elemento, se devuelve ese elemento
    if len(lista) == 1:
        return lista[0]
    else:
        # Se suma el primer elemento con la suma del resto de la lista
        return lista[0] + suma_recursiva(lista[1:])
````

````python
lista = [10, -2, 34, 78, 23, 11, -76, 132, 1]

print(f"Suma con loops: {suma_loops(lista)}")
print(f"Suma recursiva: {suma_recursiva(lista)}")
````

## Ejemplo 2: Factorial

Creen la función `factorial_loop(numero)` y `factorial_recursivo(numero)` que reciba como input un número entero positivo y devuelva su factorial:

$$ N! = N \times (N-1)\times(N-2)\times(N-3)\times \cdots \times 3\times 2\times 1$$
$$ 5! = 5 \times 4 \times 3 \times 2 \times 1=120$$

<center>
<img src="./img/recursividad.gif">
</center>

````python
def factorial_loop(numero):
    resultado = 1
    while numero > 1:
        resultado *= numero
        numero -= 1
    return resultado
````

````python
def factorial_recursivo(numero):
    # Caso base: 1! = 1
    if numero == 1:
        return numero
    # Caso recursivo: n! = n * (n-1)!
    else:
        return numero * factorial_recursivo(numero - 1)
````

````python
print(f"Factorial iterativo de 10: {factorial_loop(10)}")
print(f"Factorial recursivo de 10: {factorial_recursivo(10)}")
````

## Ejemplo 3: Exponenciación Rápida (Recursión Exitosa)

No toda recursión es igual. Veamos un caso donde la recursión es **más eficiente** que la iteración.

Si queremos calcular el interés compuesto, necesitamos elevar una base a un exponente $(1+r)^n$.
*   *Iterativo:* Multiplicar la base $n$ veces. Costo: $O(n)$.
*   *Recursivo Rápido:* Usar la propiedad matemática $x^n = (x^{n/2})^2$. Si queremos $x^{10}$, calculamos $x^5$ y lo multiplicamos por sí mismo. Cortamos el problema a la mitad en cada paso. Costo: $O(\log n)$.

Este es un ejemplo del patrón **Divide y Vencerás** aplicado con recursión.

````python
import time

def potencia_iterativa(base, exp):
    """
    Calcula base^exp multiplicando iterativamente.
    Complejidad: O(n) donde n es el exponente.
    """
    resultado = 1
    for _ in range(exp):
        resultado *= base
    return resultado

def potencia_rapida(base, exp):
    """
    Calcula base^exp usando recursión y divide y vencerás.
    Complejidad: O(log n).
    """
    # Caso base
    if exp == 0:
        return 1
    
    # Si el exponente es par: x^n = (x^(n/2))^2
    mitad = potencia_rapida(base, exp // 2)
    if exp % 2 == 0:
        return mitad * mitad
    # Si el exponente es impar: x^n = x * (x^(n//2))^2
    else:
        return base * mitad * mitad

# Comparación
base_interes = 1.05
anios = 500000

t0 = time.time()
potencia_iterativa(base_interes, anios)
t1 = time.time()
print(f"Potencia Iterativa O(n): {t1 - t0:.6f} segundos")

t0 = time.time()
potencia_rapida(base_interes, anios)
t1 = time.time()
print(f"Potencia Rápida O(log n): {t1 - t0:.6f} segundos")
````

## El Lado Oscuro de la Recursión: Fibonacci

La recursión es elegante, pero tiene un lado oscuro: si no tenemos cuidado, podemos terminar recalculando la misma información una y otra vez, destruyendo nuestra eficiencia.

> **Analogía:** Imagina que te pido calcular $10 + 10$. Me respondes $20$. Luego te pido calcular $10 + 10 + 10$. En lugar de sumar $20 + 10$, empiezas desde cero: $10 + 10 = 20$, más $10 = 30$. ¡Estás desperdiciando esfuerzo porque olvidaste el resultado anterior!

La secuencia de Fibonacci suma los dos números anteriores: $F(n) = F(n-1) + F(n-2)$. Si lo programamos ingenuamente, la computadora dibujará un árbol de llamadas que recalcula los mismos valores exponencialmente.

### Traza Visual de `fib(4)` ingenuo

```text
          fib(4)
         /      \
    fib(3)      fib(2)
    /    \      /    \
fib(2) fib(1) fib(1) fib(0)
 /   \
fib(1) fib(0)
```

¡Nota cómo `fib(2)` se calcula **dos veces**! Para `fib(40)`, se recalcularía millones de veces. El costo crece exponencialmente: $O(2^n)$.

````python
def fibonacci_ingenuo(n):
    """
    Calcula el n-ésimo número de Fibonacci de forma recursiva pura.
    Complejidad: O(2^n) - ¡Crecimiento exponencial catastrófico!
    """
    if n <= 1:
        return n
    return fibonacci_ingenuo(n - 1) + fibonacci_ingenuo(n - 2)

# Probemos con un número pequeño
print(f"Fibonacci(10) = {fibonacci_ingenuo(10)}")
print(f"Fibonacci(20) = {fibonacci_ingenuo(20)}")
````

````python
# Midamos cuánto tarda con n=35
# Advertencia: tardará varios segundos. ¡Con n=50 la PC se congelaría!
import time

t0 = time.time()
resultado = fibonacci_ingenuo(35)
t1 = time.time()
print(f"Fibonacci(35) = {resultado}")
print(f"Tiempo: {t1 - t0:.4f} segundos")
print("\n¡Imagina cuánto tardaría con n=50 o n=100!")
````

## La Solución: Memoización

La **memoización** (del inglés *memoization*) es la técnica correcta para implementar recursión cuando hay subproblemas repetidos. La idea es simple pero poderosa:

> **Regla de oro:** Si una función pura (mismos inputs = mismos outputs) es costosa, guarda su resultado en un diccionario la primera vez que la calcules. La próxima vez que necesites ese cálculo, búscalo en el diccionario en tiempo $O(1)$ en lugar de recalcularlo.

### ¿Cómo funciona?

1. Antes de calcular `fib(n)`, revisamos si ya lo calculamos antes (consultamos nuestro diccionario/memoria).
2. Si ya está en la memoria, lo devolvemos instantáneamente.
3. Si no está, lo calculamos, lo **guardamos** en la memoria, y luego lo devolvemos.

### Traza Visual de `fib(4)` con memoización

```text
          fib(4)              memoria = {}
         /      \
    fib(3)      fib(2) ← ¡YA CALCULADO! Lo busca en memoria
    /    \                   memoria = {0: 0, 1: 1, 2: 1}
fib(2)  fib(1)
 /   \
fib(1) fib(0)
```

Con memoización, cada valor de Fibonacci se calcula **exactamente una vez**. El costo pasa de $O(2^n)$ a $O(n)$.

````python
def fibonacci_memo(n, memoria=None):
    """
    Calcula Fibonacci guardando los resultados previos en un diccionario.
    Complejidad: O(n) - Crecimiento lineal.
    """
    # Inicializamos el diccionario de memoria en la primera llamada
    if memoria is None:
        memoria = {}
        
    # Si ya calculamos este valor, lo devolvemos instantáneamente
    if n in memoria:
        return memoria[n]
        
    # Casos base
    if n <= 1:
        return n
        
    # Si no lo hemos calculado, lo calculamos y lo GUARDAMOS en la memoria
    resultado = fibonacci_memo(n - 1, memoria) + fibonacci_memo(n - 2, memoria)
    memoria[n] = resultado
    
    return resultado
````

````python
# Comparación directa
import time

numero_fib = 35
print(f"Calculando Fibonacci({numero_fib})...")

t0 = time.time()
res_ingenuo = fibonacci_ingenuo(numero_fib)
t1 = time.time()
print(f"Ingenuo O(2^n): {t1 - t0:.4f} segundos (Resultado: {res_ingenuo})")

t0 = time.time()
res_memo = fibonacci_memo(numero_fib)
t1 = time.time()
print(f"Con Memoización O(n): {t1 - t0:.6f} segundos (Resultado: {res_memo})")

print(f"\n¡La versión memoizada es instantánea!")
````

## Bonus: `functools.lru_cache` — Memoización automática en Python

Python incluye una herramienta en su biblioteca estándar que hace la memoización por nosotros: el decorador `@lru_cache` del módulo `functools`.

En lugar de crear manualmente un diccionario de memoria, simplemente decoramos nuestra función con `@lru_cache` y Python se encarga de guardar y buscar los resultados automáticamente.

```python
from functools import lru_cache

@lru_cache(maxsize=None)  # maxsize=None significa caché ilimitado
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
```

El decorador `@lru_cache` convierte automáticamente cualquier función recursiva en una función memoizada. Es la forma "profesional" de implementar memoización en Python.

````python
from functools import lru_cache
import time

@lru_cache(maxsize=None)
def fibonacci_pro(n):
    """Fibonacci con memoización automática usando lru_cache."""
    if n <= 1:
        return n
    return fibonacci_pro(n - 1) + fibonacci_pro(n - 2)

t0 = time.time()
resultado = fibonacci_pro(100)
t1 = time.time()

print(f"Fibonacci(100) = {resultado}")
print(f"Tiempo: {t1 - t0:.6f} segundos")
````

## ¿Cuándo usar recursión?

La recursión no es siempre la mejor opción. Aquí una guía práctica:

| Situación | ¿Recursión? | ¿Por qué? |
| :--- | :--- | :--- |
| Problemas que se dividen naturalmente (árboles, Merge Sort) | Sí | La estructura del problema es recursiva por naturaleza. |
| Problemas con subproblemas repetidos (Fibonacci, caminos) | Sí, con memoización | Sin memoización el costo es exponencial. |
| Iteraciones simples (sumar una lista, recorrer datos) | No, usa `for` | Un loop es más claro y eficiente. |
| Problemas con muchos niveles de profundidad | Con cuidado | Python tiene un límite de ~1000 llamadas recursivas (stack overflow). |

### Reglas de Oro

1. **Siempre define un caso base claro.** Sin él, la recursión nunca termina.
2. **Asegúrate de que cada llamada recursiva avance hacia el caso base.** Si no reduces el problema, entras en un bucle infinito.
3. **Si hay subproblemas repetidos, usa memoización.** Es la diferencia entre $O(n)$ y $O(2^n)$.
4. **Si puedes resolverlo con un loop simple, prefiere el loop.** La recursión tiene un costo de memoria adicional (el call stack).
