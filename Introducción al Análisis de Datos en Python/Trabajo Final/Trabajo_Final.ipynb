## Objetivo

El notebook final debe ser un **tutorial completo y autocontenido** que le permita a cualquier persona con conocimientos básicos de Python seguir el paso a paso para resolver un problema real usando datos reales. 

Piénsenlo así: si un compañero de otra universidad descarga su notebook, ¿puede seguirlo de principio a fin sin necesitar ayuda externa? Si la respuesta es sí, van por buen camino.

## Estructura obligatoria del notebook

El notebook debe tener **exactamente estas 5 secciones**, en este orden. Cada sección tiene requisitos específicos detallados a continuación.

```
Sección 1 — Contexto y motivación
Sección 2 — Datos
Sección 3 — Metodología y desarrollo
Sección 4 — Resultados y visualización
Sección 5 — Ejercicios para el lector
```

### Sección 1 — Contexto y motivación

**¿Qué incluir?**

- **El problema:** ¿Qué problema económico o social están abordando? ¿Por qué importa?
- **La pregunta:** ¿Qué pregunta concreta buscan responder? Debe ser específica, no genérica.
- **Antecedentes:** Breve revisión de contexto. ¿Qué se ha hecho antes sobre este tema? Incluir mínimo **3 referencias** académicas o institucionales (papers, informes del DANE, Banco de la República, Banco Mundial, etc.) con enlace verificable.
- **¿Por qué Python?** Un párrafo breve explicando por qué Python (y los paquetes elegidos) son la herramienta adecuada para abordar este problema.

**Extensión mínima:** 500 palabras de texto explicativo en celdas Markdown.

**Ejemplo de lo que NO sirve:**
> "Vamos a usar scikit-learn porque es un paquete de machine learning muy popular."

**Ejemplo de lo que SÍ sirve:**
> "La predicción de la deserción escolar es un problema crítico para la política educativa en Colombia. Según X (2023), la tasa de deserción en secundaria rural fue del Y%, pero esta cifra esconde una enorme heterogeneidad entre municipios. Nuestro tutorial utiliza scikit-learn porque permite comparar múltiples modelos de clasificación (regresión logística, random forest, gradient boosting) con una interfaz unificada de fit/predict, lo cual es ideal para un análisis comparativo donde no conocemos a priori cuál modelo se ajusta mejor a los datos." 

### Sección 2 — Datos

**¿Qué incluir?**

- **Fuente:** ¿De dónde vienen los datos? Incluir URL o instrucciones de descarga exactas. Los datos deben ser **reales y de acceso público.**
- **Carga y limpieza:** Código comentado para cargar los datos, manejar valores faltantes, corregir tipos de datos, renombrar columnas, etc.
- **Descripción estadística:** Dimensiones del dataset, tipos de variables, estadísticas descriptivas básicas (`.describe()`, `.info()`, distribuciones). Acompañar con texto que interprete estos números.
- **Si combinan fuentes:** Explicar el proceso de merge/join, las llaves de unión, y verificar que no se pierdan observaciones inesperadamente.

**Requisito clave:** El proceso de obtención de datos debe ser **reproducible**. Si el lector ejecuta el notebook, debe poder obtener los mismos datos (o el notebook debe incluir instrucciones claras si se requiere descarga manual).

**Extensión mínima:** 300 palabras de texto explicativo + código funcional.

### Sección 3 — Metodología y desarrollo

Esta es la sección central del tutorial. Aquí es donde demuestran que entienden tanto la **teoría** como el **código**.

**¿Qué incluir?**

- **Explicación conceptual:** Antes de escribir código, expliquen la teoría detrás de lo que van a hacer. Si usan un modelo de regresión, expliquen qué es y qué supuestos tiene. Si usan clustering, expliquen la intuición. No se trata de copiar definiciones de Wikipedia, sino de explicarlo con sus palabras como si le enseñaran a un compañero.

- **Desarrollo paso a paso:** El código debe estar intercalado con explicaciones en Markdown. La lógica es: celda de Markdown (explico qué voy a hacer y por qué) → celda de código (lo hago) → celda de Markdown (interpreto el resultado).

- **Documentación de funciones/métodos:** Cada función o método nuevo que utilicen debe estar documentado con:
  - Qué hace
  - Qué parámetros recibe (los principales)
  - Por qué eligieron esos parámetros en su contexto específico

- **Funciones propias:** Se espera que definan al menos **2 funciones propias** (con `def`) que encapsulen lógica repetitiva o compleja. Cada función propia debe tener un *docstring* que explique qué hace.

**Requisitos:**
- Mínimo **15 funciones/métodos distintos** documentados y explicados (no solo usados, explicados).
- Al menos **2 paquetes de Python** que interactúen entre sí (ej. pandas + matplotlib, geopandas + folium, sklearn + pandas).
- **Ninguna celda de código puede superar las 30 líneas** sin tener una celda de Markdown explicativa antes. Si necesitan más de 30 líneas, dividan el código en celdas más pequeñas con explicaciones entre ellas.

**Extensión mínima:** 800 palabras de texto explicativo + código funcional.

### Sección 4 — Resultados y visualización

**¿Qué incluir?**

- Mínimo **5 visualizaciones** (gráficos, mapas, tablas formateadas con estilo).
- Cada visualización debe tener:
  - **Título** descriptivo
  - **Ejes etiquetados** (si aplica)
  - **Leyenda** (si hay múltiples series/categorías)
  - **Párrafo interpretativo** debajo que explique qué muestra la visualización y qué implica para la pregunta del tutorial
- Los resultados deben **responder a la pregunta** planteada en la Sección 1. Cierren con un párrafo de conclusión.

**Ejemplo de interpretación que NO sirve:**
> "El gráfico muestra una tendencia creciente."

**Ejemplo de interpretación que SÍ sirve:**
> "El gráfico muestra que los municipios con menor cobertura de acueducto tienen sistemáticamente menores puntajes promedio en las Pruebas Saber 11 (r = -0.67, p < 0.01). Esta relación es más pronunciada en los departamentos de Chocó, Nariño y Cauca, lo cual es consistente con la literatura sobre trampas de pobreza en regiones con baja provisión de bienes públicos (López & Núñez, 2007)." 

### Sección 5 — Ejercicios para el lector

El tutorial debe incluir **mínimo 3 ejercicios** diseñados para que alguien que haya seguido el tutorial pueda practicar por su cuenta.

**Cada ejercicio debe tener:**

1. **Enunciado claro:** ¿Qué se le pide al lector que haga?
2. **Nivel de dificultad:** Fácil / Intermedio / Difícil
3. **Pistas o hints:** Al menos una pista para guiar al lector
4. **Solución:** En una celda de código al final del notebook (pueden usar un encabezado "## Soluciones" al final)

**Los ejercicios deben ser progresivos:**
- **Ejercicio fácil:** Replicar algo hecho en el tutorial con una pequeña variación (ej. cambiar la variable dependiente, filtrar por otro departamento).
- **Ejercicio intermedio:** Combinar dos o más conceptos del tutorial (ej. limpiar un nuevo dataset y aplicar el mismo modelo).
- **Ejercicio difícil:** Extender el análisis de una forma no cubierta en el tutorial (ej. agregar una nueva fuente de datos, probar una técnica diferente).

## Consejos finales

1. **Empiecen por los datos.** El 80% de los problemas de un proyecto de datos están en conseguir, limpiar y entender los datos. No subestimen esta parte.

2. **No intenten abarcar demasiado.** Es mejor un tutorial profundo y bien explicado sobre un tema acotado que un tutorial superficial que toca muchos temas.

3. **Prueben que su notebook corre en limpio.** Antes de entregar, hagan `Kernel → Restart & Run All` y verifiquen que no hay errores. Esto es un requisito mínimo y es la forma más fácil de perder puntos.

4. **Piensen en su lector.** El mejor tutorial es el que ustedes hubieran querido encontrar cuando estaban aprendiendo. Escriban para esa persona.

5. **Revisen la checklist.** Los requisitos mínimos están diseñados para ser objetivos y verificables. Antes de entregar, revísenlos uno por uno. Es la diferencia entre poder aspirar a 5.0 o tener techo en 3.0.

6. **La IA es una herramienta, no el autor.** Pueden usar ChatGPT, Copilot o lo que quieran para aprender, generar código inicial, o resolver dudas. Pero el tutorial debe reflejar su comprensión del tema, sus datos, y su criterio. Un notebook que se lee como output directo de un LLM se nota, y se evalúa en consecuencia.
