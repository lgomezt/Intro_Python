## ¿Qué es Pandas?

Pandas es una librería de código abierto para Python que proporciona herramientas de análisis de datos de alto rendimiento y fáciles de usar. Está construida sobre la biblioteca NumPy y proporciona estructuras de datos de alto nivel para trabajar con datos estructurados y tabulares, como hojas de cálculo y bases de datos relacionales.

Algunas de las características más importantes de Pandas son:

- **Estructuras de datos poderosas**: Ofrece el `DataFrame` (tabla bidimensional, similar a una hoja de Excel) y la `Series` (columna individual de datos). Estas son las dos estructuras fundamentales con las que trabajaremos.
- **Lectura y escritura de datos**: Permite importar y exportar datos desde y hacia una variedad de formatos de archivo, como CSV, Excel, SQL, JSON, entre otros.
- **Limpieza y preprocesamiento**: Facilita operaciones como la eliminación de valores perdidos, la transformación de tipos de datos y la normalización de datos.
- **Análisis y manipulación**: Ofrece métodos para agrupar, agregar, filtrar, ordenar y transformar datos de manera eficiente.
- **Integración con otras librerías**: Se integra con bibliotecas de visualización como Matplotlib y Seaborn para crear gráficos a partir de los datos.

Pandas es ampliamente utilizado en ciencia de datos, finanzas, ingeniería y otras disciplinas que requieren la manipulación y el análisis de grandes conjuntos de datos. Es una de las herramientas más populares en Python para el análisis de datos debido a su facilidad de uso y su amplia gama de funciones.

En el siguiente diagrama se puede ver la estructura general de un `DataFrame`. Es muy similar a una hoja de Excel: las **filas** representan observaciones (individuos, registros) y las **columnas** representan variables (características, atributos). Cada celda contiene un valor específico.

### Creación de un DataFrame desde un diccionario

Comenzaremos nuestra introducción a Pandas creando nuestro primer `DataFrame` a partir de un diccionario de Python. Esta es una de las formas más intuitivas de crear un `DataFrame`, ya que la estructura de un diccionario se mapea de forma natural a una tabla:

- Cada **llave** del diccionario se convierte en el **nombre de una columna**.
- Cada **valor** (que debe ser una lista) se convierte en los **datos de esa columna**.

> **Importante:** Todas las listas deben tener la misma longitud, de lo contrario Pandas generará un error.

````python
# Primero importamos la librería pandas.
# La convención estándar es importarla con el alias "pd" para escribir menos código.
import pandas as pd
````

````python
# Revisemos la versión de pandas de este tutorial
pd.__version__
````

````python
# Cada variable de nuestro dataframe partirá de una lista.
# Definimos una lista por cada columna que queremos crear.
nombres = ["Jose", "Sebastian", "Hamadys", "Carlos", "Julian", "Juan"]
sexo = ["M", "M", "F", "M", "M", None]  # None representa un valor faltante
edad = [24, 25, 32, 45, 52, 20]
peso = [70, 80, 55, 90, 75, 65]  # Peso en kilogramos
````

````python
# Creamos un diccionario donde las llaves serán los nombres de las columnas
# y los valores serán las listas que acabamos de definir.
un_diccionario = {
    "names": nombres,
    "sex": sexo,
    "age": edad,
    "weight": peso
}
````

````python
un_diccionario
````

````python
# Construimos nuestro primer DataFrame pasando el diccionario a pd.DataFrame()
# Pandas automáticamente convierte cada llave en una columna y cada lista en sus valores.
data = pd.DataFrame(un_diccionario)
````

````python
# Visualizamos nuestro dataframe
data
````

````python
# Podemos guardar nuestro DataFrame como un archivo de Excel.
# Más adelante veremos en detalle cómo exportar datos en distintos formatos.
# index=False evita que se exporte el índice numérico como una columna adicional.
data.to_excel("Datos/data_estudiantes.xlsx", index = False)
````

### Indexación y ejes

Para manipular DataFrames es fundamental aprender a **extraer porciones de nuestra información**. A esto se le conoce generalmente como *slicing* (rebanado). Piense en ello como recortar una hoja de Excel para quedarse solo con las filas o columnas que necesita.

En esta sección aprenderemos los principales métodos de selección y manipulación:

| Método | Descripción |
|--------|-------------|
| `df["columna"]` | Extraer una o más columnas |
| `.set_index()` | Definir una columna como índice |
| `.reset_index()` | Reiniciar el índice a valores numéricos |
| `.loc[]` | Seleccionar por **etiquetas** (nombres) de filas y columnas |
| `.iloc[]` | Seleccionar por **posición numérica** (enteros) |
| `.sort_values()` | Ordenar las filas por los valores de una columna |
| `.rename()` | Cambiar el nombre de filas o columnas |
| `.unique()` | Ver los valores únicos de una columna |
| `.value_counts()` | Contar la frecuencia de cada valor único |

#### Extraer columnas

Para extraer columnas de un DataFrame tenemos dos métodos:

1. **Corchetes `[]`**: Escribimos el nombre de la columna entre comillas dentro de corchetes cuadrados. Si queremos extraer **varias columnas**, pasamos una lista de nombres.
2. **Notación de punto `.`**: Escribimos el nombre de la columna después de un punto. Este método es más rápido de escribir, pero **solo funciona** si el nombre de la columna no tiene espacios ni caracteres especiales, y no coincide con un método de Pandas.

> **Nota:** Al extraer una sola columna, el resultado es una `Series` (estructura unidimensional). Al extraer varias, el resultado es un `DataFrame`.

````python
# Método 1: Corchetes con el nombre de la columna.
# El resultado es una Series (una sola columna con su índice).
data["names"]
````

````python
# Para extraer múltiples columnas, pasamos una lista de nombres dentro de los corchetes.
# El resultado es un DataFrame (subconjunto de la tabla original).
data[["names", "weight"]]
````

````python
# También podemos guardar los nombres de columnas en una variable
# y pasarla como argumento. Esto es útil cuando la selección es dinámica.
columnas_deseadas = ["names", "weight"]
data[columnas_deseadas]
````

````python
# Método 2: Notación de punto.
# Más cómodo, pero no funciona con nombres que tengan espacios o caracteres especiales.
data.names
````

#### El índice del DataFrame

Antes de continuar, es importante entender qué es el **índice** de un DataFrame. El índice es la columna de etiquetas que identifica cada fila. Por defecto, Pandas asigna un índice numérico (0, 1, 2, ...), pero podemos definir cualquier columna como índice usando `.set_index()`.

**¿Por qué cambiar el índice?**
- Facilita la búsqueda de datos específicos (ej: buscar por nombre).
- Permite usar `.loc[]` con etiquetas significativas.
- Mejora el rendimiento en búsquedas frecuentes.

> **Importante sobre `inplace`:** Muchos métodos de Pandas no modifican el DataFrame original, sino que devuelven una **copia modificada**. Para guardar el cambio, podemos:
> 1. Reasignar: `data = data.set_index('names')`
> 2. Usar el parámetro `inplace=True`: `data.set_index('names', inplace=True)`

````python
# .set_index() convierte una columna en el índice del DataFrame.
# Aquí usamos 'names' como índice. Note que NO estamos guardando el resultado.
data.set_index('names')
````

````python
# Como no reasignamos el resultado (data = ...), el DataFrame original no cambió.
# Esta es una confusión muy común para principiantes.
data 
````

````python
# Ahora sí guardamos el cambio reasignando la variable.
data = data.set_index('names')
data
````

````python
# Otra forma: usar inplace=True modifica el DataFrame directamente sin necesidad de reasignar.
# Aquí agregamos "sex" como un segundo nivel de índice (MultiIndex) con append=True.
data.set_index("sex", append = True, inplace = True)
data
````

````python
# .reset_index() deshace el set_index, devolviendo las columnas del índice a columnas normales.
data.reset_index(inplace = True)
data
````

````python
# Volvemos a definir "names" como índice para los siguientes ejemplos.
data.set_index("names", inplace = True)
````

````python
data
````

#### Filtrado con `.loc[]` y `.iloc[]`

Ahora que sabemos extraer columnas y manejar el índice, pasemos a los dos métodos más poderosos de selección en Pandas: `.loc` y `.iloc`. La diferencia clave es:

| Método | Selección por | Ejemplo |
|--------|--------------|---------|
| `.loc[]` | **Etiquetas** (nombres) | `data.loc["Jose", "age"]` |
| `.iloc[]` | **Posición numérica** (enteros) | `data.iloc[0, 1]` |

Ambos usan la sintaxis `[filas, columnas]`, donde se puede pasar un valor individual, una lista de valores, o un *slice* (rango).

Los métodos `.iloc` y ``.loc`` son utilizados en Pandas para acceder a los datos de un dataframe de diferentes maneras.

`.iloc` se utiliza para acceder a los datos del dataframe mediante índices enteros. Permite seleccionar filas y columnas utilizando números enteros en lugar de etiquetas de fila y columna. La sintaxis general para utilizar `.iloc` es la siguiente:

```python
dataframe.iloc[filas, columnas]
```

Donde `filas` y `columnas` son listas o arrays de números enteros que indican las filas y columnas que se desean seleccionar, respectivamente.

Por ejemplo, para seleccionar las primeras 3 filas y las primeras 2 columnas de un dataframe, se puede utilizar el siguiente código:

```python 
dataframe.iloc[:3, :2]
```

Recuerde que la indexación en python empieza en 0. A continuación puede encontrar una referencia gráfica:

Por otro lado, `.loc` se utiliza para acceder a los datos del dataframe mediante etiquetas de fila y columna. Permite seleccionar filas y columnas utilizando las etiquetas de fila y columna en lugar de índices enteros. La sintaxis general para utilizar `.loc` es la siguiente:

```python
dataframe.loc[filas, columnas]
```

Donde `filas` y `columnas` son listas o arrays de etiquetas de fila y columna que indican las filas y columnas que se desean seleccionar, respectivamente.

<center>
<div>
<img src="img/iloc3.png" width="500"/>
</div>
</center>

Por ejemplo, para seleccionar todas las filas donde la columna "age" es mayor o igual a 30, se puede utilizar el siguiente código:

```python
dataframe.loc[dataframe['age'] >= 30, :]
```

En resumen, `.iloc` se utiliza para acceder a los datos mediante índices enteros, mientras que `.loc` se utiliza para acceder a los datos mediante etiquetas de fila y columna. Ambos métodos son muy útiles para la selección y manipulación de datos en Pandas.

````python
# Extraiga la fila de Hamadys usando .loc
data.loc['Hamadys', :]
````

````python
# .loc[] usa etiquetas. Como definimos "names" como índice, buscamos por nombre.
# La coma separa filas de columnas. ":" significa "todas".
data.loc['Hamadys', :]
````

````python
# Extraiga la columna de edad usando .loc
data.loc[:,'age']
````

````python
# Extraiga los pesos de Juan y Carlos usando .loc
data.loc[['Juan', 'Carlos'], 'weight']
````

````python
# Extraiga a todas las personas de sexo masculino mayores a 30 años
data.loc[(data['age'] > 30) & (data['sex'] == 'M'),]
````

````python
# Filtrado booleano: una de las características más poderosas de Pandas.
# Podemos combinar condiciones usando:
#   & (AND): ambas condiciones deben cumplirse
#   | (OR): al menos una condición debe cumplirse
#   ~ (NOT): niega la condición
# IMPORTANTE: cada condición individual debe ir entre paréntesis.
data.loc[(data['age'] > 30) & (data['sex'] == 'M'),]
````

````python
# Extraiga la segunda fila usando .iloc
data.iloc[1]
````

````python
data.iloc[1,]
````

````python
# Extraiga la segunda y tercera columna usando .iloc
data.iloc[:,[1,2]]
````

````python
# Extraiga la primera y última fila y segunda columna usando .iloc
data.iloc[[0,-1],[1]]
````

````python
# Reinicie el índice
data.reset_index(inplace = True, drop = False) 
data
````

### Modificación de valores

Hasta ahora hemos aprendido a explorar y seleccionar datos. Ahora veremos cómo **modificar** el contenido de un DataFrame. Esto incluye:
- Editar valores individuales (celdas específicas).
- Modificar columnas completas.
- Crear nuevas columnas.

Las columnas dentro de un DataFrame son objetos llamados `Series`. Una `Series` es una estructura de datos **unidimensional** que contiene una secuencia de valores con un índice. Puede pensarla como una columna individual de una hoja de cálculo.

````python
# Verifiquemos: al extraer una columna, obtenemos un objeto de tipo Series.
type(data["names"])
````

#### Editar valores individuales

Podemos modificar valores específicos combinando `.loc[]` con una asignación. Esto es especialmente útil para **corregir datos erróneos o faltantes**. Por ejemplo, el sexo de Juan está registrado como nulo (`None`). Vamos a corregirlo.

````python
# Primero verificamos el valor actual: efectivamente es None (nulo).
data.loc[data.names == "Juan", "sex"]
````

````python
# Ahora lo corregimos asignándole el valor "M" (masculino).
# Usamos .loc[] para ubicar la celda exacta y luego el operador = para asignar.
data.loc[data.names == "Juan", "sex"] = "M"
````

````python
data
````

#### Modificar columnas completas

También podemos aplicar operaciones a una columna entera. Esto es útil para **transformar unidades**, aplicar fórmulas, o normalizar datos. Actualmente tenemos la edad en años; vamos a convertirla a días multiplicando por 365.

````python
# Multiplicamos toda la columna "age" por 365.
# La operación se aplica elemento por elemento a cada valor de la columna.
data["age"] = data["age"]*365
````

````python
data
````

#### Crear nuevas columnas

Frecuentemente necesitamos agregar nuevas variables a nuestro DataFrame. Hay dos formas principales de hacerlo:

1. **Asignación directa**: `data["nueva_columna"] = valores`
2. **Método `.assign()`**: `data = data.assign(nueva_columna = valores)`

Ambos métodos logran lo mismo. La diferencia es que `.assign()` retorna una copia del DataFrame (útil para encadenar operaciones), mientras que la asignación directa modifica el DataFrame original.

````python
# Forma 1: Asignación directa.
# Creamos una columna llena de valores nulos (np.nan = "Not a Number").
import numpy as np
data["nueva_variable"] = np.nan
````

````python
data
````

Usando `.assign()` logramos lo mismo pero de una forma que permite encadenar múltiples operaciones:

````python
data = data.assign(nueva_variable2 = np.nan)
data
````

#### Modificar con valores diferentes por fila

El valor con el que se modifica cada fila no tiene que ser el mismo para todas. Podemos pasar una **lista** con un valor distinto para cada fila. La lista debe tener la misma longitud que el número de filas del DataFrame.

````python
apellido = ["Urrutia", "Carvajal", "Gómez", "Tobón", "Pardo", "López"]
````

````python
data["names"] = data["names"] + " " + apellido
````

### Estadísticas descriptivas

Anteriormente vimos cómo `.describe()` proporciona un resumen estadístico general. Ahora veremos cada función estadística por separado, lo que nos da más control sobre qué calculamos y cómo.

Las funciones más comunes son:

| Función | Descripción |
|---------|-------------|
| `.sum()` | Suma de los valores |
| `.min()` / `.max()` | Valor mínimo / máximo |
| `.idxmin()` / `.idxmax()` | Índice del valor mínimo / máximo |
| `.mean()` | Media (promedio) aritmética |
| `.median()` | Mediana (valor central) |
| `.std()` | Desviación estándar |

> **Recuerde:** el parámetro `axis` controla la dirección del cálculo. `axis=0` (por defecto) calcula por columna; `axis=1` calcula por fila.

````python
# .sum() suma los valores de cada columna.
# Para columnas de texto, concatena los strings (¡cuidado con esto!).
data.sum()
````

````python
# Con axis=1, .sum() suma los valores de cada fila.
# numeric_only=True ignora las columnas no numéricas para evitar errores.
data.sum(axis = 1, numeric_only = True)
````

````python
# También podemos aplicar funciones a una columna específica.
data["weight"].sum()
````

````python
# .min() devuelve el valor mínimo de cada columna.
data.min()
````

````python
# .max() devuelve el valor máximo de cada columna.
data.max()
````

````python
# .idxmin() devuelve el ÍNDICE (posición) del valor mínimo.
# Esto es útil para identificar qué observación tiene el valor más bajo.
data.age.idxmin()
````

````python
# Combinando .idxmin() con .iloc[] podemos extraer la fila completa
# de la persona con la edad mínima.
data.iloc[data.age.idxmin(),]
````

````python
# ¿Cuál es el índice de la persona con mayor peso?
data.weight.idxmax()
````

### Aplicar funciones a las columnas
En Pandas, el método `apply` se utiliza para aplicar una función a una serie o a un dataframe. El método `apply` toma como argumento una función y la aplica a cada elemento de la serie o del dataframe, devolviendo una nueva serie o dataframe con los resultados de la función aplicada a cada elemento.

La función que se aplica en el método `apply` puede ser una función lambda o una función definida por el usuario. Esta función debe tomar como argumento un elemento de la serie o dataframe y devolver el resultado de aplicar la función al elemento.

El método `apply` se puede aplicar de varias formas en un dataframe:

- Aplicar una función a una columna específica: Se puede utilizar el método `apply` en una columna específica de un dataframe utilizando la siguiente sintaxis:

```python
dataframe["nombre_columna"].apply(funcion)
```

Donde "nombre_columna" es el nombre de la columna en la que se desea aplicar la función y "funcion" es la función que se desea aplicar a cada elemento de la columna.

- Aplicar una función a un dataframe completo: Se puede utilizar el método `apply` en un dataframe completo utilizando la siguiente sintaxis:

```python
dataframe.apply(funcion)
```

Donde "funcion" es la función que se desea aplicar a cada elemento del dataframe.

- Aplicar una función a una fila específica: Se puede utilizar el método `apply` en una fila específica de un dataframe utilizando la siguiente sintaxis:

```python
dataframe.loc[etiqueta_fila].apply(funcion)
```

Donde "etiqueta_fila" es la etiqueta de la fila en la que se desea aplicar la función y "funcion" es la función que se desea aplicar a cada elemento de la fila.

- el método `apply` también puede ser utilizado con funciones **lambda**, que son funciones anónimas y pequeñas que se definen dentro de una línea de código. Por ejemplo, si se quisiera aplicar una función **lambda** que tome como entrada una cadena de caracteres y devuelva la versión en mayúsculas de la misma, se podría utilizar el método apply de la siguiente manera:

```python
dataframe['nombre_columna'].apply(lambda x: x.upper())
```

En resumen, el método `apply` se utiliza en Pandas para aplicar una función a una serie o dataframe, y puede utilizarse de diferentes maneras para aplicar la función.

### Concatenar DataFrames con `pd.concat`

La función `pd.concat()` se utiliza para **unir dos o más DataFrames** a lo largo de un eje. Piense en ello como "pegar" tablas:
- **`axis=0`** (por defecto): Pega verticalmente (una tabla debajo de otra). Útil cuando las tablas tienen las **mismas columnas**.
- **`axis=1`**: Pega horizontalmente (una tabla al lado de otra). Útil cuando las tablas tienen las **mismas filas**.

```python
pd.concat([df1, df2], axis=0)   # Vertical: apilar filas
pd.concat([df1, df2], axis=1)   # Horizontal: unir columnas
```

Los parámetros más importantes son:
- `axis`: Dirección de la concatenación (0 = vertical, 1 = horizontal).
- `join`: Tipo de unión (`'outer'` incluye todo, `'inner'` solo lo común).
- `ignore_index`: Si es `True`, reinicia el índice del resultado.

> **Nota:** En la próxima clase veremos `pd.merge()`, que es más poderoso para unir DataFrames por columnas en común (similar a un JOIN en SQL).

````python
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df1
````

````python
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df2
````

````python
nuevo_df = pd.concat([df1, df2], axis=0)
nuevo_df
````

En este caso, se utilizará `axis=0` para concatenar los dataframes verticalmente. Los índices de los dataframes originales se conservarán en el nuevo dataframe.

La función `pd.concat` también puede ser utilizada para combinar dataframes o series que no tienen columnas o índices en común. En este caso, las columnas o índices faltantes se rellenarán con valores nulos. Por ejemplo, si se quisiera combinar horizontalmente dos dataframes que tienen diferentes columnas, se podría utilizar el siguiente código:

````python
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df1
````

````python
df2 = pd.DataFrame({'C': [5, 6], 'D': [7, 8]})
df2
````

````python
nuevo_df = pd.concat([df1, df2], axis=1)
nuevo_df
````

En este caso, se utilizará `axis=1` para concatenar los dataframes horizontalmente. Como los dataframes no tienen columnas en común, se crearán nuevas columnas con valores nulos.

En resumen, la función `pd.concat` en Pandas es una herramienta muy útil para combinar dataframes o series a lo largo de un eje específico. Se puede utilizar con diferentes parámetros para personalizar la unión de los objetos y crear un nuevo dataframe o serie.

````python
# Creemos una nueva fila
nueva_fila = pd.DataFrame({"names": "Andrea Saldarriaga", "sex": "F", "age": 12000, "weight": 60}, index = [0])
data = pd.concat((data, nueva_fila), axis = 0)
data
````

### Importar datos con Pandas

En la práctica, casi nunca creamos DataFrames desde cero. Lo más común es **importar datos** desde archivos existentes. Pandas soporta una gran variedad de formatos:

#### CSV (Comma-Separated Values)
El formato más común y universal para datos tabulares. Es un archivo de texto donde cada fila es una línea y las columnas se separan por comas (u otro delimitador).

```python
# Lectura básica
df = pd.read_csv('archivo.csv')

# Con separador diferente (ej: punto y coma, común en archivos de Excel en español)
df = pd.read_csv('archivo.csv', sep=';')

# Especificar codificación (útil para caracteres especiales como ñ, tildes)
df = pd.read_csv('archivo.csv', encoding='utf-8')

# Usar una columna como índice
df = pd.read_csv('archivo.csv', index_col='ID')
```

#### Excel
Pandas puede leer archivos `.xlsx` y `.xls` directamente.

```python
# Lectura básica (lee la primera hoja por defecto)
df = pd.read_excel('archivo.xlsx')

# Leer una hoja específica
df = pd.read_excel('archivo.xlsx', sheet_name='Hoja2')
```

#### Pickle
Formato binario de Python para guardar objetos. Es más rápido que CSV/Excel y preserva los tipos de datos exactamente como están.

```python
# Guardar
df.to_pickle('archivo.pkl')

# Leer
df = pd.read_pickle('archivo.pkl')
```

#### Parámetros comunes de lectura

| Parámetro | Descripción |
|-----------|-------------|
| `sep` | Carácter separador (por defecto `,` en CSV) |
| `header` | Fila que contiene los nombres de columnas |
| `index_col` | Columna para usar como índice |
| `usecols` | Subconjunto de columnas a leer |
| `skiprows` | Filas a saltar al inicio del archivo |
| `encoding` | Codificación del archivo (`utf-8`, `latin-1`, etc.) |
