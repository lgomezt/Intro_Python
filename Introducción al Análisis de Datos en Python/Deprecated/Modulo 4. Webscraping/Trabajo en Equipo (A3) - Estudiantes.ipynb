# Trabajo en Equipo (A3)
## Webscraping
1. Visite la página web de Wikipedia acerca de los [Oscars a Mejor Película](https://es.wikipedia.org/wiki/Anexo:Óscar_a_la_mejor_película).
2. Encuentra la sección que dice **"Películas ganadoras y candidatas"**.
3. Cree un código para scrapear la página usando un requests.
4. Construya una tabla que consolide la información desde 1920 hasta 2024. Esta tabla debe tener las columnas: "Título original", "Título en español", "Compañía productora", "Productor(es)", "Director(es)", "Género(s)", "Número de nominaciones", "Número de premios", "Candidatas", "Ceremonia".

    Ceremonia debe ser por ejemplo: "1.ª ceremonia de entrega de los Premios Óscar (1927/28)"

5. Repita el web scraping pero esta vez usando Selenium.
