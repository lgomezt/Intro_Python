6. Cree un código que juegue "Piedra, papel y tijera" con el usuario. La idea es que al usuario se le pida su elección entre "piedra", "papel" o "tijera" y luego la máquina haga una elección aleatoria entre las 3 opciones. Luego usted deberá comunicar el resultado del juego.

```python
import numpy as np

mi_eleccion = "Tijera"
computador = np.random.choice(["Piedra", "Papel", "Tijera"])
print(f"Elegiste {mi_eleccion}, el computador eligió {computador}")
print("")
if mi_eleccion == computador:
    print("Empate! ambos eligieron", mi_eleccion)
elif mi_eleccion == "Piedra":
    if computador == "Tijera":
        print("Ganaste! Piedra rompe Tijera")
    elif computador == "Papel":
        print("Perdiste! Papel envuelve Piedra")
elif mi_eleccion == "Papel":
    if computador == "Piedra":
        print("Ganaste! Papel envuelve Piedra")
    elif computador == "Tijera":
        print("Perdiste! Tijera corta Papel")
elif mi_eleccion == "Tijera":
    if computador == "Papel":
        print("Ganaste! Tijera corta Papel")
    elif computador == "Piedra":
        print("Perdiste! Piedra rompe Tijera")
```
