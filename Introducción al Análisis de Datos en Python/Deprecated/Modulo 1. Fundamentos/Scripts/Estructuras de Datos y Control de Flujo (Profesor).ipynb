# Parte 1: Estructuras de Datos

Python ofrece varias formas de organizar datos. Las tres más importantes son **listas**, **diccionarios** y **tuplas**. Cada una tiene características distintas que las hacen útiles en diferentes situaciones.

## 1.1 Listas

Una **lista** es una colección ordenada de elementos que puede contener cualquier tipo de dato. Se define con corchetes `[]` y los elementos se separan por comas.

Características clave:
- **Ordenada**: los elementos mantienen el orden en que los colocamos
- **Mutable**: podemos agregar, eliminar o cambiar elementos después de crearla
- **Indexada desde 0**: el primer elemento está en la posición 0, no en la 1

Piensen en una lista como una columna de datos en una hoja de cálculo.

````python
# Tasas de crecimiento del PIB (%) de países latinoamericanos en 2023
paises = ["Colombia", "México", "Chile", "Perú", "Argentina", "Brasil"]
crecimiento_pib = [1.2, 3.2, 0.2, -0.6, -1.6, 2.9]

print(paises)
print(crecimiento_pib)
````

### Indexación y slicing

Para acceder a elementos de una lista usamos su **índice** (posición). Recuerden: Python empieza a contar desde 0.

```
  Índices:     0          1        2       3          4          5
            ["Colombia", "México", "Chile", "Perú", "Argentina", "Brasil"]
  Negativos: -6         -5       -4      -3         -2         -1
```

````python
# Acceder a un elemento por su posición
print(paises[0])    # Primer elemento
print(paises[2])    # Tercer elemento
print(paises[-1])   # Último elemento
````

````python
# Slicing: obtener un subconjunto de la lista
# Sintaxis: lista[inicio:fin]  (incluye inicio, NO incluye fin)
print(paises[1:4])   # Del índice 1 al 3
print(paises[:3])    # Desde el inicio hasta el índice 2
print(paises[3:])    # Desde el índice 3 hasta el final
````

### Métodos esenciales de listas

Las listas tienen **métodos** (funciones incorporadas) que nos permiten manipularlas:

````python
# .append() - Agregar un elemento al final
paises.append("Ecuador")
crecimiento_pib.append(2.4)
print(paises)
````

````python
# .remove() - Eliminar un elemento por su valor
paises.remove("Ecuador")
crecimiento_pib.remove(2.4)
print(paises)
````

````python
# .pop() - Eliminar y devolver un elemento por su posición
ultimo = paises.pop()  # Sin argumento, elimina el último
print(f"Se eliminó: {ultimo}")
print(paises)
````

````python
# Volvamos a agregar Brasil
paises.append("Brasil")

# .sort() - Ordenar la lista (modifica la lista original)
paises_ordenados = paises.copy()  # Hacemos una copia para no perder el orden original
paises_ordenados.sort()
print(paises_ordenados)
````

````python
# Concatenar listas con +
paises_andinos = ["Colombia", "Perú", "Ecuador"]
paises_cono_sur = ["Chile", "Argentina", "Uruguay"]
todos = paises_andinos + paises_cono_sur
print(todos)
````

### Funciones útiles con listas

Python tiene funciones integradas que trabajan directamente con listas numéricas:

````python
crecimiento_pib = [1.2, 3.2, 0.2, -0.6, -1.6, 2.9]

print(f"Número de países: {len(crecimiento_pib)}")
print(f"Suma total: {sum(crecimiento_pib)}")
print(f"Máximo crecimiento: {max(crecimiento_pib)}")
print(f"Mínimo crecimiento: {min(crecimiento_pib)}")

# Calcular el promedio
promedio = sum(crecimiento_pib) / len(crecimiento_pib)
print(f"Crecimiento promedio: {round(promedio, 2)}%")
````

### Mutabilidad: las listas se pueden modificar

````python
# Podemos cambiar cualquier elemento directamente
paises = ["Colombia", "México", "Chile", "Perú", "Argentina", "Brasil"]
print(f"Antes: {paises[1]}")

paises[1] = "Panamá"
print(f"Después: {paises[1]}")
print(paises)
````

## 1.2 Diccionarios

Un **diccionario** almacena datos como pares de **llave-valor**. Se define con llaves `{}`.

Características clave:
- **Pares llave:valor**: cada dato tiene una etiqueta (llave) que lo identifica
- **Mutable**: podemos agregar, eliminar o cambiar elementos
- **Acceso por llave**: en vez de buscar por posición (como en listas), buscamos por nombre

**¿Cuándo usar un diccionario en vez de una lista?** Cuando tus datos tienen **nombres o etiquetas naturales**. Por ejemplo, los indicadores económicos de un país tienen nombre (PIB, inflación, desempleo), entonces un diccionario es más claro que una lista.

````python
# Indicadores económicos de Colombia (2023)
colombia = {
    "pib_billones_usd": 334.7,
    "inflacion": 9.28,
    "desempleo": 10.2,
    "poblacion_millones": 52.2,
    "moneda": "Peso colombiano"
}

print(colombia)
````

````python
# Acceder a un valor usando su llave
print(f"PIB de Colombia: {colombia['pib_billones_usd']} billones USD")
print(f"Inflación: {colombia['inflacion']}%")
````

````python
# Agregar una nueva llave
colombia["tasa_interes"] = 13.25
print(colombia)
````

````python
# Modificar un valor existente
colombia["inflacion"] = 8.35
print(f"Inflación actualizada: {colombia['inflacion']}%")
````

````python
# Métodos útiles de diccionarios
print("Llaves:", list(colombia.keys()))     # Todas las llaves
print("Valores:", list(colombia.values()))   # Todos los valores
````

### Diccionarios anidados

Los valores de un diccionario pueden ser cualquier cosa: números, strings, listas, e incluso **otros diccionarios**. Esto es muy útil para representar datos estructurados.

````python
# Datos económicos de varios países
latam = {
    "Colombia": {"pib": 334.7, "inflacion": 9.28, "desempleo": 10.2},
    "México":   {"pib": 1322.7, "inflacion": 4.66, "desempleo": 2.8},
    "Chile":    {"pib": 335.5, "inflacion": 7.56, "desempleo": 8.5},
    "Brasil":   {"pib": 2173.7, "inflacion": 4.62, "desempleo": 7.8}
}

# Acceder al PIB de México
print(f"PIB de México: {latam['México']['pib']} billones USD")
````

````python
# Un diccionario con listas como valores: portafolio de inversión
portafolio = {
    "Ecopetrol":   {"precio": 2150, "acciones": 100},
    "Bancolombia": {"precio": 34500, "acciones": 20},
    "ISA":         {"precio": 18900, "acciones": 50}
}

# Valor de la posición en Ecopetrol
valor_ecopetrol = portafolio["Ecopetrol"]["precio"] * portafolio["Ecopetrol"]["acciones"]
print(f"Valor en Ecopetrol: ${valor_ecopetrol:,} COP")
````

## 1.3 Tuplas

Una **tupla** es similar a una lista, pero con una diferencia fundamental: es **inmutable**. Una vez creada, no se puede modificar.

Se define con paréntesis `()`.

**¿Cuándo usar tuplas?** Cuando tienes datos que **no deben cambiar** por naturaleza: coordenadas geográficas, fechas de referencia, categorías fijas.

````python
# Coordenadas geográficas de Bogotá (no deberían cambiar)
coordenadas_bogota = (4.6097, -74.0817)
print(f"Latitud: {coordenadas_bogota[0]}, Longitud: {coordenadas_bogota[1]}")

# Trimestres del año (son fijos)
trimestres = ("Q1", "Q2", "Q3", "Q4")
print(trimestres)
````

### Desempaquetado (unpacking)

Una característica muy útil de las tuplas es que podemos extraer sus valores en variables separadas de un solo golpe:

````python
# Desempaquetado: asignar cada elemento a una variable
registro = (2024, "Q1", 320.5)
anio, trimestre, pib = registro

print(f"Año: {anio}")
print(f"Trimestre: {trimestre}")
print(f"PIB: {pib} billones USD")
````

### Inmutabilidad: las tuplas NO se pueden modificar

````python
# Esto genera un error (y está bien que lo haga)
try:
    coordenadas_bogota[0] = 5.0
except TypeError as e:
    print(f"Error: {e}")
    print("Las tuplas no se pueden modificar. Esa es su ventaja: protegen datos que no deben cambiar.")
````

### Resumen: ¿Cuándo usar qué?

| Estructura | Sintaxis | Mutable | Caso de uso |
|:---|:---:|:---:|:---|
| **Lista** | `[a, b, c]` | Sí | Colección de datos que puede crecer o cambiar |
| **Diccionario** | `{"a": 1, "b": 2}` | Sí | Datos con etiquetas o nombres (llave-valor) |
| **Tupla** | `(a, b, c)` | No | Datos fijos que no deben modificarse |

**Regla práctica:** Si necesitan almacenar datos y van a manipularlos, usen una **lista** o un **diccionario**. Si los datos son fijos y constantes, usen una **tupla**.

# Parte 2: Control de Flujo

Hasta ahora sabemos cómo almacenar datos. Pero un programa útil necesita **tomar decisiones** y **repetir acciones**. Para eso existen las estructuras de control de flujo:

- **Condicionales** (`if`/`elif`/`else`): ejecutar código solo si se cumple una condición
- **Loops** (`for`/`while`): repetir código múltiples veces

## 2.1 Condicionales: `if`, `elif`, `else`

Los condicionales permiten que el programa ejecute diferentes bloques de código dependiendo de si una condición es **verdadera** (`True`) o **falsa** (`False`).

**Sintaxis:**
```python
if condición:
    # código si es True
elif otra_condición:
    # código si la anterior fue False y esta es True
else:
    # código si todas las anteriores fueron False
```

**Operadores de comparación:**
| Operador | Significado |
|:---:|:---|
| `==` | Igual a |
| `!=` | Diferente de |
| `>` | Mayor que |
| `<` | Menor que |
| `>=` | Mayor o igual que |
| `<=` | Menor o igual que |

**Operadores lógicos:** `and` (y), `or` (o), `not` (negación)

> **Importante:** El bloque de código dentro del `if` debe estar **indentado** (con 4 espacios o un tab). Así es como Python sabe qué código pertenece al condicional.

````python
# Caso base: if simple
inflacion = 9.28

if inflacion > 5:
    print("La inflación es alta")
````

````python
# if/else: dos caminos posibles
tasa_desempleo = 3.5

if tasa_desempleo > 8:
    print("Desempleo elevado: se necesitan políticas de empleo")
else:
    print("Desempleo en niveles manejables")
````

````python
# if/elif/else: múltiples categorías
# Clasificación de países por PIB per cápita según el Banco Mundial
pib_per_capita = 6630  # Colombia, aprox. USD

if pib_per_capita >= 14006:
    clasificacion = "Ingreso alto"
elif pib_per_capita >= 4516:
    clasificacion = "Ingreso medio-alto"
elif pib_per_capita >= 1146:
    clasificacion = "Ingreso medio-bajo"
else:
    clasificacion = "Ingreso bajo"

print(f"Con un PIB per cápita de ${pib_per_capita:,} USD, el país es de: {clasificacion}")
````

````python
# Combinando condiciones con 'and' y 'or'
inflacion = 12.5
desempleo = 15.3

if inflacion > 10 and desempleo > 12:
    print("Alerta: indicadores de posible crisis económica")
elif inflacion > 10 or desempleo > 12:
    print("Precaución: al menos un indicador está en niveles críticos")
else:
    print("Indicadores dentro de rangos aceptables")
````

> **Nota:** Un `if` puede existir solo. Los `elif` y `else` siempre necesitan un `if` previo. Pueden haber múltiples `elif`, pero solo un `else` (al final).

## 2.2 Loops: `for`

Un **`for` loop** repite un bloque de código **una vez por cada elemento** de una secuencia (lista, tupla, diccionario, o rango de números).

**Sintaxis:**
```python
for elemento in secuencia:
    # código que se ejecuta en cada iteración
```

Es la herramienta ideal cuando sabemos **sobre qué datos** queremos iterar.

````python
# Caso base: iterar sobre una lista
paises = ["Colombia", "México", "Chile", "Perú"]

for pais in paises:
    print(pais)
````

### `range()`: generar secuencias numéricas

La función `range()` genera una secuencia de números. Es muy útil cuando queremos repetir algo un número específico de veces.

- `range(n)`: genera números de 0 a n-1
- `range(inicio, fin)`: genera números de inicio a fin-1
- `range(inicio, fin, paso)`: genera números con un incremento específico

````python
# range() básico
for i in range(5):
    print(i)
````

````python
# range() con inicio, fin y paso
# Años de censos económicos (cada 5 años)
for anio in range(2000, 2026, 5):
    print(f"Censo económico del año {anio}")
````

### Loop + condicional: donde se pone interesante

La verdadera potencia aparece al combinar loops con condicionales:

````python
# Clasificar países según su tasa de crecimiento
paises = ["Colombia", "México", "Chile", "Perú", "Argentina", "Brasil"]
crecimiento_pib = [1.2, 3.2, 0.2, -0.6, -1.6, 2.9]

for i in range(len(paises)):
    if crecimiento_pib[i] > 2:
        estado = "crecimiento fuerte"
    elif crecimiento_pib[i] > 0:
        estado = "crecimiento moderado"
    else:
        estado = "recesión"
    print(f"{paises[i]}: {crecimiento_pib[i]}% -> {estado}")
````

### Iterar sobre diccionarios

Con `.items()` podemos iterar sobre las llaves y valores de un diccionario al mismo tiempo:

````python
indicadores_colombia = {
    "PIB (billones USD)": 334.7,
    "Inflación (%)": 9.28,
    "Desempleo (%)": 10.2,
    "Tasa de interés (%)": 13.25
}

for indicador, valor in indicadores_colombia.items():
    print(f"{indicador}: {valor}")
````

### Acumular resultados con un loop

Un patrón muy común: empezar con un valor inicial e irlo actualizando en cada iteración.

````python
# Calcular el PIB total de la región
latam = {
    "Colombia": {"pib": 334.7, "inflacion": 9.28, "desempleo": 10.2},
    "México":   {"pib": 1322.7, "inflacion": 4.66, "desempleo": 2.8},
    "Chile":    {"pib": 335.5, "inflacion": 7.56, "desempleo": 8.5},
    "Brasil":   {"pib": 2173.7, "inflacion": 4.62, "desempleo": 7.8}
}

pib_total = 0
for pais, datos in latam.items():
    pib_total = pib_total + datos["pib"]

promedio_pib = pib_total / len(latam)
print(f"PIB total de la región: {pib_total} billones USD")
print(f"PIB promedio: {round(promedio_pib, 1)} billones USD")
````

### Construir listas con un loop

Otro patrón fundamental: empezar con una lista vacía e ir agregándole elementos.

````python
# Extraer solo los países con inflación mayor al 5%
paises_alta_inflacion = []

for pais, datos in latam.items():
    if datos["inflacion"] > 5:
        paises_alta_inflacion.append(pais)

print(f"Países con inflación > 5%: {paises_alta_inflacion}")
````

### List comprehension: la versión compacta

Python ofrece una forma concisa de crear listas a partir de otras secuencias, llamada **list comprehension**:

```python
nueva_lista = [expresion for variable in secuencia]
nueva_lista = [expresion for variable in secuencia if condicion]
```

Es exactamente lo mismo que el patrón de lista vacía + for + append, pero en una sola línea.

````python
# Sin list comprehension (lo que ya sabemos)
cuadrados = []
for x in [1, 2, 3, 4, 5]:
    cuadrados.append(x ** 2)
print(cuadrados)

# Con list comprehension (misma lógica, una línea)
cuadrados = [x ** 2 for x in [1, 2, 3, 4, 5]]
print(cuadrados)
````

````python
# List comprehension con condición
crecimiento_pib = [1.2, 3.2, 0.2, -0.6, -1.6, 2.9]

# Solo los valores positivos
positivos = [tasa for tasa in crecimiento_pib if tasa > 0]
print(f"Tasas positivas: {positivos}")
````

> **¿Cuándo usar list comprehension?** Cuando la operación es sencilla (1-2 líneas de lógica). Si es más compleja, un `for` explícito es más legible.

## 2.3 Loops: `while`

Un **`while` loop** repite un bloque de código **mientras una condición sea verdadera**.

**Sintaxis:**
```python
while condicion:
    # código que se ejecuta mientras condicion sea True
```

**¿Cuándo usar `while` en vez de `for`?**
- `for`: cuando sabes **sobre qué** iterar (una lista, un rango, un diccionario)
- `while`: cuando **no sabes cuántas veces** necesitas repetir, solo sabes **cuándo parar**

````python
# Caso base
contador = 0
while contador < 5:
    print(contador)
    contador += 1  # Equivale a: contador = contador + 1
````

````python
# Ejemplo económico: ¿en cuántos años se duplica una deuda con interés compuesto?
deuda_inicial = 1000  # USD
tasa_interes = 0.08   # 8% anual
deuda_actual = deuda_inicial
anios = 0

while deuda_actual < 2 * deuda_inicial:
    deuda_actual = deuda_actual * (1 + tasa_interes)
    anios += 1
    print(f"Año {anios}: deuda = ${deuda_actual:,.2f}")

print(f"\nLa deuda se duplica en {anios} años")
````

> **Sobre `break` y `continue`:** Dentro de cualquier loop pueden usarse `break` (salir del loop inmediatamente) y `continue` (saltar a la siguiente iteración). Son útiles pero en la práctica la mayoría de situaciones se resuelven bien con condicionales dentro del loop.

# Parte 3: Ejemplo Integrador

Combinemos todo lo aprendido en un ejercicio práctico. Supongamos que somos analistas en un organismo internacional y tenemos datos económicos de varios países de América Latina.

````python
# Datos económicos de América Latina
datos_economicos = {
    "Colombia":  {"pib_pc": 6630,  "inflacion": 9.28,  "desempleo": 10.2, "region": "Andina"},
    "México":    {"pib_pc": 10820, "inflacion": 4.66,  "desempleo": 2.8,  "region": "Norteamérica"},
    "Chile":     {"pib_pc": 16800, "inflacion": 7.56,  "desempleo": 8.5,  "region": "Cono Sur"},
    "Perú":      {"pib_pc": 7000,  "inflacion": 3.24,  "desempleo": 6.2,  "region": "Andina"},
    "Argentina": {"pib_pc": 13650, "inflacion": 133.5, "desempleo": 6.4,  "region": "Cono Sur"},
    "Brasil":    {"pib_pc": 10400, "inflacion": 4.62,  "desempleo": 7.8,  "region": "Atlántico"},
    "Ecuador":   {"pib_pc": 6280,  "inflacion": 2.22,  "desempleo": 3.5,  "region": "Andina"},
    "Uruguay":   {"pib_pc": 21580, "inflacion": 5.11,  "desempleo": 8.3,  "region": "Cono Sur"}
}
````

### Tarea 1: Clasificar países por nivel de ingreso

````python
# Clasificar cada país según umbrales del Banco Mundial
for pais, datos in datos_economicos.items():
    pib_pc = datos["pib_pc"]
    
    if pib_pc >= 14006:
        nivel = "Ingreso alto"
    elif pib_pc >= 4516:
        nivel = "Ingreso medio-alto"
    elif pib_pc >= 1146:
        nivel = "Ingreso medio-bajo"
    else:
        nivel = "Ingreso bajo"
    
    print(f"{pais}: ${pib_pc:,} USD -> {nivel}")
````

### Tarea 2: Calcular la inflación promedio de los países andinos

````python
suma_inflacion = 0
cantidad = 0

for pais, datos in datos_economicos.items():
    if datos["region"] == "Andina":
        suma_inflacion += datos["inflacion"]
        cantidad += 1
        print(f"  {pais}: {datos['inflacion']}%")

promedio = suma_inflacion / cantidad
print(f"\nInflación promedio de países andinos: {round(promedio, 2)}%")
````

### Tarea 3: Encontrar el país con mayor PIB per cápita

````python
mayor_pib = 0
pais_mayor = ""

for pais, datos in datos_economicos.items():
    if datos["pib_pc"] > mayor_pib:
        mayor_pib = datos["pib_pc"]
        pais_mayor = pais

print(f"El país con mayor PIB per cápita es {pais_mayor} con ${mayor_pib:,} USD")
````

### Bonus: Tarea 2 con list comprehension

````python
# La misma tarea 2 pero de forma compacta
inflaciones_andinas = [datos["inflacion"] for pais, datos in datos_economicos.items() if datos["region"] == "Andina"]

promedio = sum(inflaciones_andinas) / len(inflaciones_andinas)
print(f"Inflaciones andinas: {inflaciones_andinas}")
print(f"Promedio: {round(promedio, 2)}%")
````
